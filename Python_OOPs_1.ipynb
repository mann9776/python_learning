### Types of Variables
- In a class, we have 3 types variables
      - Instance variable (object level variable)
      - Static variable (class level variable)
      - Local variable (method level variable)
### Types of methods in a class
- In python we have 3 different types of methods:
   - Instance method
  - Static Method
  - Class Method

```python
class Student:
    def __init__(self):
        self.name = 'Manmohan'
        self.roll_no = 101
        self.age = 33
        self.marks = 90
    def talk(self):
        print(f'I am {self.name}')
        print(f'My roll_no is {self.roll_no}')
        print(f'My age is {self.age}')
        print(f'I have got {self.marks}')
        
        
```

### Constructor

- A constructor is a special method in python
- The name of the constructor is fixed i.e., __init__()
- Whenever we create an object, the contructor is automatically executed.
- The main objective of a constructor is to declare an instance (object-level variable)
- The constructor gets executed only once for each object when the object is created.
- A constructor must have atleast an argument (self)
- A contructor is optional

```python
class Student:
    def __init__(self):
        print('Contructor Executed...')
        self.name = 'Manmohan' #instance variable
        self.roll_no = 101 #Instance variable
        self.age = 33 #Instance variable
        self.marks = 90
    def talk(self):
        print(f'I am {self.name}')
        print(f'My roll_no is {self.roll_no}')
        print(f'My age is {self.age}')
        print(f'I have got {self.marks}')
```

```python
s = Student()
```

```python
s.talk()
```

```python
class Student:
    def __init__(self, a, b, c, d):
        print('Contructor Executed...')
        self.name = a #instance variable
        self.roll_no = b #Instance variable
        self.age = c #Instance variable
        self.marks = d #Instance variable
    def talk(self):
        print(f'I am {self.name}')
        print(f'My roll_no is {self.roll_no}')
        print(f'My age is {self.age}')
        print(f'I have got {self.marks}')
```

```python
s1 = Student('Abha', 123, 22, 88)
```

```python
s1.talk()
```

### Self
- Self is an implicit variable provided by python that always points to the current object
- Using self variable, we access the instance variable and the methods.

```python
class Test:
    def __init__(self):
        pass
    def greet():
        pass
```

```python
t = Test()
```

```python
t.greet()
```

```python
class Test:
    def __init__(self):
        pass
    def greet(self):
        pass
```

```python
t1 = Test()
```

```python
t1.greet()
```

```python
class Test:
    def __init__(self, a, b):
        self.name = a
        self.age = b
    def my_name(self):
        print(f'I am {self.name}')        
```

```python
t = Test('Prabha', 25)
```

```python
t.my_name()
```
