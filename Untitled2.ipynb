### Static Variable (Class level variable)

```python
class Customer:
    def __init__(self, name, amt):
        self.bankname = 'SBI'
        self.name = name
        self.amount = amt
    def display(self):
        print(f'Welcome to {self.bankname}')
        print(f'The customer is {self.name}')
        print(f'The amount in user\'s account - {self.amount}')
```

```python
t = Customer('Kuldeep', 25000)
```

```python
t.display()
```

### How to declare and initialise a static variable?
- Generally we declare a static variable inside the class before any method
- Inside an instance method by using a class name
- Inside a class method by using class name or cls variable
- Inside the static method by using class name
- Outside the class by using class name

```python
### Case-2: Inside an instance method by using a class name
class Customer:
    def __init__(self):
        pass
    def method(self):
        Customer.bank = 'SBI'
```

```python
t1 = Customer()
```

```python
t1.method()
```

```python
Customer.__dict__
```

### Case - 3: Inside a class method by using class name or cls variable

```python
class Customer:
    def __init__(self):
        pass
    @classmethod
    def cls_method(cls):
        cls.a = 'SBI'
        cls.b = 'RBI'
    
```

```python
t = Customer()
```

```python
t.cls_method()
```

```python
Customer.__dict__
```

### Case 4: Inside the static method by using class name

```python
class Customer:
    def __init__(self):
        pass
    @staticmethod
    def stat():
        Customer.bank = 'RBI'
```

```python
t = Customer()
```

```python
t.stat()
```

```python
Customer.__dict__
```

### Case 5: Outside the class by using class name

```python
Customer.x = 'HDFC'
```

```python
Customer.__dict__
```

#### WAP to create a class Movie, name, hero, heroine, rating, display

```python
class Movie:
    def __init__(self, name, hero, heroine, rating):
        self.movie_name = name
        self.movie_hero = hero
        self.movie_heroine = heroine
        self.movie_rating = rating
        self.display()
    def display(self):
        print(f'Name of the movie is {self.movie_name}')
        print(f'{self.movie_hero} and {self.movie_heroine} performed in this movie')
        print(f'rating of the movie is {self.movie_rating} star')
    
```

- Single Inheritance
- Hierarchical Inheritance

```python
class P:
    def m1(self):
        print('Parent Class method')
class C1(P):
    def m2(self):
        print('C1 class method')
class C2(P):
    def m3(self):
        print('C2 class method')
class C3(P):
    def m4(self):
        print('C3 class method')
```
