```python
def add(x,y):
    return x+y

def sub(x,y):
    return x-y

def factorial(n):
    return 1 if (n==1 or n==0) else n * factorial(n - 1)

def mul(x,y):
    return x*y

def div(x,y):
    if y == 0:
        print('Error: Division with 0 is not possible')
    else return x/y

def power(x,y):
    return x**y

def floor_div(x,y):
    if y == 0:
        print('Error: Division with 0 is not possible')
    else return x//y

def cumulative_sum(*a):
    sum = 0
    for i in a:
        sum += i
    return sum
    
```
